```python
from sklearn import linear_model
from sklearn import datasets
import pandas as pd
import matplotlib.pyplot as plot
import numpy as np

# all columns
allCols = pd.read_csv("data/lifestyle.csv")
allCols = allCols.iloc[:, 1:-3]
numCols = len(allCols.columns)

# fruits and veggies 

# stress levels
y = allCols.iloc[:, 1]
X = allCols.iloc[:, 0]
corrCoef = np.corrcoef(X, y)
print('corrCoef: ', corrCoef)

X = X.values.reshape(-1, 1)

lm = linear_model.LinearRegression()
model = lm.fit(X, y)

plot.scatter(X, y, color='red', alpha=.1)
plot.plot(X, lm.predict(X), color='blue')
plot.ylabel('Stress Level (Rated 1-5)')
plot.xlabel('Number of Fruits/Vegetables Consumed Daily')
plot.title('Effect of Fruit and Vegetable Consumption on Stress Levels')
plot.show()


```
